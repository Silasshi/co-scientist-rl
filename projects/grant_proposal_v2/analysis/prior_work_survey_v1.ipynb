## Per-paper breakdown — is this applicable to grant proposal?

### 1. DR Tulu / RLER (Ai2 2025) — HIGH THREAT

- **Output**: Deep research reports with citation tags (no length spec, but long-form)
- **Rubric evolution**: rubric buffer, bounded size, drops near-zero-variance rubrics. Timing periodic during RL, not per-step
- **Rubric gen**: search-grounded (web + paper search seeds rubric generator)
- **Mechanism**: GRPO-style loss + positive rubrics (new strategies) + negative rubrics (penalize hacking)
- **Goodhart**: negative rubrics explicitly penalize 'copying retrieved text verbatim, padding, overusing citations'
- **No critique-revise in training loop**
- **Grant proposal**: NOT mentioned. No NSF/NIH/ERC.

**Overlap with GER-CR-v1**: evolving rubric buffer with std-filter is directly parallel to your `filter_and_truncate`. The pattern 'drop dead-channel rubrics' is in DR Tulu too.

**Your differentiator**: (a) grant proposal domain with public agency rubric (they use auto-gen), (b) evolving rubric + CR combined (they have only evolving rubric), (c) dead-channel taxonomy with 3 attack surfaces.

---

### 2. Query-Specific Rubrics for DeepResearch (2602.03619, Feb 2026) — HIGH THREAT

- **Output**: DeepResearch long-form reports
- **Method**: train a query-specific rubric GENERATOR from 5000 pref pairs via GRPO; hybrid reward = preference consistency + format + LLM-as-a-judge
- **Rubric gen**: learned generator, not fixed
- **Evolving**: per-query (not per-training-iter)
- **Grant proposal**: NOT mentioned. Benchmarked on DeepResearch Bench (100 queries).

**Overlap with GER-CR-v1**: training rubric generator is a different architecture from your R_active binary grader, but the spiritual claim ('learned rubric > fixed rubric') overlaps.

**Your differentiator**: you use public agency rubric as persistent anchor + evolving layer on top. They have no anchor — full learned.

---

### 3. Goel 2512 — Training AI Co-Scientists Using Rubric Rewards (Meta + Oxford + Max Planck, Dec 2025) — MEDIUM

- **Output**: Research plan, 750 words in `<solution>` tags. NOT grant proposal.
- **Method**: GRPO with frozen initial policy as reward model. Rubrics auto-extracted from papers (Llama-4-Maverick candidates, Claude-4-Sonnet filter, top 10 per goal).
- **Single signal**: fraction of satisfied rubric items + format penalty
- **Results**: 70% expert win rate (ML domain), cross-domain transfer (med → ML +15%)
- **Grant proposal**: NOT mentioned
- **No critique-revise**

**This is the paper you already know (memory has it as 'Goel 2512, 70% expert preference')**.

**Your differentiator**: they do research plan, you do grant proposal. They have no CR, no evolving rubric, no attack-surface analysis. D3 work directly competes with them; D4 has domain escape.

---

### 4. Rubric-ARM / Alternating RL (2602.01511, Feb 2026) — MEDIUM

- **Domain**: General open-ended post-training
- **Method**: alternating train rubric generator + judge; EM-like coupling via shared correctness objective
- **Rubric gen**: query-specific, evolving via alternation
- **Grant proposal**: NOT mentioned
- **No critique-revise**

**Overlap with GER-CR-v1**: the 'alternate rubric and policy' idea is adjacent to your Phase 3. But they alternate rubric generator and judge, you alternate rubric buffer and policy. Different components.

**Your differentiator**: you have CR + external anchor + Goodhart taxonomy.

---

### 5. RuscaRL — Rubric-Scaffolded RL (2508.16949) — LOW

- **Domain**: General reasoning (med, creative, IF, STEM)
- **Method**: 2-phase (exploration with rubric scaffolding / exploitation with rubric reward). Rubric usage decays via sigmoid.
- **Rubric**: HealthBench-like checklist; GPT-4.1 generates non-medical ones
- **Evolving**: NO. Rubric fixed, usage decays.
- **Grant proposal**: NOT mentioned

**Your differentiator**: full domain + method + evolution + taxonomy mismatch. Can cite as background not as direct competitor.

### 6. RIFL — Rubric-Based Instruction Following (Meta 2511.10507) — LOW

- **Method**: rubric gen + finetuned verifier + reward shaping
- **Result**: +6.7% on AdvancedIF
- **Grant proposal**: NOT mentioned

**Relevance**: methodology adjacent but domain is short-form IF.

### 7. Rethinking RL-Based Deep Research Agents (2510.15862) — NONE

### 8. ResearchRubrics benchmark (2511.07685) — YOU MUST REPORT SCORES

- **Benchmark only** (no training method proposed)
- **Scale**: 2800 human-hours, 2500+ expert rubrics, 3 dims (factual grounding / reasoning soundness / clarity)
- **State-of-the-art (as of Nov 2025)**: Gemini DR and OpenAI DR both < 68% compliance
- **Scope**: long-form deep research answers. Not grant proposals specifically, but closest public benchmark.

**Implication**: reviewers will ask 'does GER-CR-v1 beat Gemini DR on ResearchRubrics?'. You need at least a row in your results table comparing to the 68% baseline, even if your main eval is a custom grant benchmark.

## What niche is left for D4v2 / GER-CR-v1

### ✅ Grant proposal domain

Zero of 8 papers does grant proposal with agency rubrics. Your Open Grants + NSF + NIH + ERC dataset is unique.

**Moat argument**: other papers use auto-extracted rubrics (Goel, DR Tulu) or learned-from-preference rubrics (Query-Specific). These are all internal to the training loop and risk circular reasoning. Public agency rubrics are externally validated and break the self-grading loop.

### ✅ Critique-revise + evolving rubric combined

| Method | CR | Evolving rubric |
|---|:-:|:-:|
| DR Tulu | ❌ | ✅ |
| Goel 2512 | ❌ | ❌ |
| Query-Specific | ❌ | per-query |
| Rubric-ARM | ❌ | ✅ |
| Self-Refine / Reflexion / SCoRe | ✅ | ❌ |
| **GER-CR-v1** | ✅ | ✅ |

You are the first to combine both. Claim: CR alone plateaus (validated: 17/40 on Phase 2 smoke); evolving rubric alone lacks targeted locus (validated: DR Tulu only hits 41.8 on DeepResearch Bench even with RLER).

### ✅ Goodhart attack-surface taxonomy

Unique to you:
- **Length bias**: ρ(len, G12)=0.78-0.82 across v8 runs; iter-0 baseline, not RL artifact
- **Judge gap**: GPT-OSS vs Opus rank inversion at iter 10
- **Signal saturation via dead channels**: G4/G10 var<0.25 on all v8 runs; 22% weight goes dead

DR Tulu says 'we use negative rubrics' — one-line acknowledgment, no taxonomy.  
Query-Specific balances 3 rewards — pragmatic, not diagnostic.  
Nobody else does this level of attack-surface analysis.

### ⚠️ scores_only = full_critique negative finding

Your claim (B4 scores_only Opus 19 = B4 full_critique Opus 19 on FoundOpt) directly challenges:
- RLTF 2602.02482 ('text feedback is intermediate signal richer than scalar')
- Self-Refine / SCoRe assumption (critique text carries information)

But this is currently n=2 condition. For a paper claim you need cross-goal + multi-seed verification.

**Working title candidates**:
- 'Grant Proposal Generation Reveals Three Goodhart Modes in Rubric-RL, and How Evolving Rubrics Close Them'
- 'Writing NSF-Compliant Proposals: A Case Study in Rubric-RL Failure Modes and Targeted Defenses'

**Frame**:
1. Grant proposals with public agency rubrics as empirical testbed (domain moat)

3. GER-CR-v1 as targeted defense per surface (methodological contribution)
4. Baselines: DR Tulu, Goel 2512, Query-Specific Rubrics on grant benchmark; also report on ResearchRubrics to compare to Gemini DR

### Option B: Negative finding first

'Critique Text is Redundant for Rubric-Guided Revision (When Scores Suffice)'

Directly challenges Self-Refine / SCoRe / Reflexion / RLTF. Risk: single-negative-finding paper is hard at NeurIPS.

### Option C: Evolving-rubric + CR combined (method novelty)

Risk: overlaps heavily with DR Tulu's RLER framing. Reviewers may not see CR addition as enough delta.

**My recommendation**: Option A. Your domain (grant proposals) + your data (3-surface taxonomy across v7/v8 runs) are the least contested parts of your work.
